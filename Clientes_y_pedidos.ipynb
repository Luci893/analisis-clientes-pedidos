**Pregunta 4** ¿Cuál es el estado con menos clientes?

```python
import csv

archivo = "./sample_data/customers.csv"

# Crear diccionario
conteo = {}
# abrir archivo e iniciar vars
with open(archivo, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')

  next(csvreader) # Salto encabezado

# evaluar el contenido del archivo
  for fila in csvreader:
    # strip() permite sacar espacios
    # upper() transforma a mayuscula
    estado = fila[4].strip().upper()
    if estado in conteo:
      conteo[estado] += 1
    else:
      conteo[estado] = 1

# Menor valor de clientes en estado
minimo = min(conteo.values())

# Encontrar los estados con ese valor minimo ya que son varios iguales
estados_menorClientes = []
for estado, valor in conteo.items():
  if valor == minimo:
    estados_menorClientes.append(estado)

print(f'Del siguiente diccionario {conteo}\nLos estados con menor clientes son = {estados_menorClientes}')
```

Pregunta 5: ¿Cuál es el apellido más común?

```python
import csv

archivo = "./sample_data/customers.csv"

# Crear diccionario
conteoAp = {}
# abrir archivo e iniciar vars
with open(archivo, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')

  next(csvreader) # Salto encabezado

# evaluar el contenido del archivo
  for fila in csvreader:
    # strip() permite sacar espacios
    # upper() transforma a mayuscula
    apellido = fila[2].strip().upper()
    if apellido in conteoAp:
      conteoAp[apellido] += 1
    else:
      conteoAp[apellido] = 1

apellidoFrecuente = max(conteoAp, key = conteoAp.get)
valorApFrecuente = max(conteoAp.values())

print(f'Del siguiente diccionario {conteoAp}\nEl apellido más frecuente es: {apellidoFrecuente} con {valorApFrecuente} repeticiones')
```

Pregunta n.º 2: ¿Cuál es el número promedio de artículos por pedido (redondeado a dos decimales)?

```python
import pandas as pd

df_orders = pd.read_csv('./sample_data/orders.csv')

# Acá agrupamos los mismos pedidos y contamos las filas en las que esta ese pedido
articulosPedido = df_orders.groupby('OrderID').size()

# Saco promedio y lo redondeo a 2 decimales
promedio = articulosPedido.mean().round(2)

print(f'El promedio de articulos por pedido es: {promedio}')
```

Pregunta #3: ¿Cuál es la mayor cantidad de artículos por pedido?

```python
import pandas as pd

df_orders = pd.read_csv('./sample_data/orders.csv')

# Agrupamos los mismos pedidos con los productos que tiene
articulosPedido = df_orders.groupby('OrderID').size()

# Saco el pedido con más articulos
maxArticulos = articulosPedido.max()
pedidoMaxArticulos = articulosPedido.idxmax()

print(f'La mayor cantidad de articulos lo tiene el pedido {pedidoMaxArticulos} con un total de {maxArticulos} articulos.')
```

Pregunta #4: ¿Cuál es la cantidad de pedidos realizados en octubre de 2021?

```python
import pandas as pd

df_orders = pd.read_csv('./sample_data/orders.csv')

# Convertir a fecha
df_orders['Date'] = pd.to_datetime(df_orders['Date'])

# Filtrar los pedidos por el mes de octubre de 2021
df_PedidosOct = df_orders[
    (df_orders['Date'] >= '2021-10-01') & (df_orders['Date'] < '2021-11-01')]

# Contar cantidad de pedidos únicos
cantidadPedidosOct = df_PedidosOct['OrderID'].nunique()

print(f'La cantidad de pedidos realizados en octubre de 2021 es de: {cantidadPedidosOct}')
```

Pregunta n.º 5: ¿Qué cliente gastó la mayor cantidad de dinero en 2021?

```python
import pandas as pd

df_orders = pd.read_csv('./sample_data/orders.csv')

import pandas as pd

df_orders = pd.read_csv('./sample_data/orders.csv')

# Convertir a fecha
df_orders['Date'] = pd.to_datetime(df_orders['Date'])

# Extraer solo fecha de la columna 'Date'
df_2021 = df_orders[df_orders['Date'].dt.year == 2021]

# Agrupamos por cliente
df_Cliente = df_2021.groupby('CustomerID')

# Sumamos la cantidad de dinero de cada cliente
df_TotalDinero = df_Cliente['OrderTotal'].sum()

# Saco el cliente que mayor dinero gasto
clienteMayorDinero = df_TotalDinero.idxmax()
dineroMayor = df_TotalDinero.max()

print(f'El cliente {clienteMayorDinero} es el que gasto más dinero en 2021 con un total de {dineroMayor}')
```

Pregunta n.º 6: Históricamente, ¿cuál es el mejor mes para las ventas?

```python
import pandas as pd

df_orders = pd.read_csv('./sample_data/orders.csv')

# Convertir a fecha
df_orders['Date'] = pd.to_datetime(df_orders['Date'])

# Extraer mes
df_orders['Mes'] = df_orders['Date'].dt.month

# Por cada pedido tomamos el valor y el mes del mismo. Sacando duplicados.
# agg() agrupa varias operaciones
df_pedidos = df_orders.groupby('OrderID').agg(
    {'OrderTotal': 'sum',
     'Mes': 'first'})

# Agrupar por mes y suma
ventas_por_mes = df_pedidos.groupby('Mes')['OrderTotal'].sum()

# Diccionario para identificar el nombre del mes al enviar el numero del mismo.
meses = {
    1: 'Enero', 2: 'Febrero', 3: 'Marzo',
    4: 'Abril', 5: 'Mayo', 6: 'Junio',
    7: 'Julio', 8: 'Agosto', 9: 'Septiembre',
    10: 'Octubre', 11: 'Noviembre', 12: 'Diciembre'
}
# Encontrar el mejor mes y su cantidad de ventas
mejor_mes = ventas_por_mes.idxmax()
mejor_mes_nom = meses[mejor_mes]
cantidad_ventas = ventas_por_mes.max()

print(f'El mejor mes para las ventas fue {mejor_mes_nom} y la cantidad de ventas fue de {cantidad_ventas}')
```
